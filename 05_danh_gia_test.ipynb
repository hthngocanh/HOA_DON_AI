```python

import time
import cv2
import pytesseract

from io import BytesIO
from PIL import Image

# ==========================================
# 1. CẤU HÌNH
# ==========================================

# Sử dụng PROJECT_DIR đã khai báo ở đầu Notebook
DATA_DIR = PROJECT_DIR / "data" / "ICDAR2019-SROIE" / "data"

TEST_PATH = DATA_DIR / "test-00000-of-00001.parquet"

OUTPUT_DIR = PROJECT_DIR / "outputs" / "ocr"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

OCR_CONFIG = "--oem 3 --psm 6"

# ==========================================
# 2. ĐỌC ẢNH TEST
# ==========================================

ocr_image_table = pq.read_table(
    TEST_PATH,
    columns=["key", "image"]
)

ocr_image_records = ocr_image_table.to_pylist()

print("Tổng số ảnh Test:", len(ocr_image_records))
print("Tesseract version:", pytesseract.get_tesseract_version())

# ==========================================
# 3. TIỀN XỬ LÝ ẢNH
# ==========================================

def preprocess_for_ocr(image):

    image_array = np.array(image.convert("RGB"))

    gray = cv2.cvtColor(
        image_array,
        cv2.COLOR_RGB2GRAY
    )

    resized = cv2.resize(
        gray,
        None,
        fx=2,
        fy=2,
        interpolation=cv2.INTER_CUBIC
    )

    return resized

# ==========================================
# 4. CHẠY TESSERACT OCR
# ==========================================

test_ocr_results = []

start_time = time.time()

for index, record in enumerate(ocr_image_records, start=1):

    invoice_key = str(record["key"])

    try:
        image_bytes = record["image"]["bytes"]

        image = Image.open(
            BytesIO(image_bytes)
        ).convert("RGB")

        processed_image = preprocess_for_ocr(image)

        recognized_text = pytesseract.image_to_string(
            processed_image,
            lang="eng",
            config=OCR_CONFIG
        )

        test_ocr_results.append({
            "key": invoice_key,
            "split": "test",
            "ocr_text": recognized_text,
            "status": "success"
        })

    except Exception as error:

        test_ocr_results.append({
            "key": invoice_key,
            "split": "test",
            "ocr_text": "",
            "status": "failed",
            "error": str(error)
        })

    if index % 25 == 0 or index == len(ocr_image_records):
        print(f"Processed {index}/{len(ocr_image_records)} invoices")

# ==========================================
# 5. KIỂM TRA VÀ LƯU KẾT QUẢ
# ==========================================

test_ocr_df = pd.DataFrame(test_ocr_results)

elapsed_time = time.time() - start_time

TEST_OCR_PATH = OUTPUT_DIR / "test_ocr.csv"

test_ocr_df.to_csv(
    TEST_OCR_PATH,
    index=False,
    encoding="utf-8-sig"
)

print("\n========== TEST OCR COMPLETED ==========")
print("Tổng số hóa đơn:", len(test_ocr_df))
print("Số mã duy nhất:", test_ocr_df["key"].nunique())
print("\nTrạng thái OCR:")
print(test_ocr_df["status"].value_counts(dropna=False))

print("\nSố kết quả OCR rỗng:")
print(
    test_ocr_df["ocr_text"]
    .fillna("")
    .str.strip()
    .eq("")
    .sum()
)

print("\nThời gian chạy:", round(elapsed_time, 2), "giây")
print("File đã lưu:", TEST_OCR_PATH)

display(test_ocr_df.head(3))

# Kiểm tra tính hợp lệ
assert len(test_ocr_df) == 361
assert test_ocr_df["key"].nunique() == 361
assert set(test_ocr_df["split"].unique()) == {"test"}

print("\nTEST OCR DATA VALIDATION COMPLETED!")
```

```python

import time
import cv2
import numpy as np
import pandas as pd
import pytesseract

from io import BytesIO
from PIL import Image
from pathlib import Path
import pyarrow.parquet as pq

# ==========================================
# 1. ĐỌC LẠI DỮ LIỆU TEST
# ==========================================

DATA_DIR = PROJECT_DIR / "data" / "ICDAR2019-SROIE" / "data"
TEST_PATH = DATA_DIR / "test-00000-of-00001.parquet"

OUTPUT_DIR = PROJECT_DIR / "outputs" / "ocr"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

ocr_image_table = pq.read_table(
    TEST_PATH,
    columns=["key", "image"]
)

ocr_image_records = ocr_image_table.to_pylist()

# ==========================================
# 2. GIỮ LẠI 281 KẾT QUẢ ĐÃ CÓ
# ==========================================

if "test_ocr_results" not in globals():
    raise NameError("Không tìm thấy kết quả OCR cũ!")

existing_results = list(test_ocr_results)

processed_keys = {
    str(item["key"])
    for item in existing_results
}

remaining_records = [
    record for record in ocr_image_records
    if str(record["key"]) not in processed_keys
]

print("Kết quả đã có:", len(processed_keys))
print("Số hóa đơn cần OCR tiếp:", len(remaining_records))

# ==========================================
# 3. HÀM TIỀN XỬ LÝ ẢNH
# ==========================================

def preprocess_for_ocr(image):

    image_array = np.array(image.convert("RGB"))

    gray = cv2.cvtColor(
        image_array,
        cv2.COLOR_RGB2GRAY
    )

    resized = cv2.resize(
        gray,
        None,
        fx=2,
        fy=2,
        interpolation=cv2.INTER_CUBIC
    )

    return resized

# ==========================================
# 4. OCR TIẾP CÁC HÓA ĐƠN CÒN THIẾU
# ==========================================

OCR_CONFIG = "--oem 3 --psm 6"

test_ocr_results = existing_results.copy()

start_time = time.time()

for index, record in enumerate(remaining_records, start=1):

    invoice_key = str(record["key"])

    try:
        image_bytes = record["image"]["bytes"]

        image = Image.open(
            BytesIO(image_bytes)
        ).convert("RGB")

        processed_image = preprocess_for_ocr(image)

        recognized_text = pytesseract.image_to_string(
            processed_image,
            lang="eng",
            config=OCR_CONFIG
        )

        test_ocr_results.append({
            "key": invoice_key,
            "split": "test",
            "ocr_text": recognized_text,
            "status": "success"
        })

    except Exception as error:

        test_ocr_results.append({
            "key": invoice_key,
            "split": "test",
            "ocr_text": "",
            "status": "failed",
            "error": str(error)
        })

    # Lưu tiến độ sau mỗi 10 hóa đơn
    if index % 10 == 0 or index == len(remaining_records):

        partial_df = pd.DataFrame(test_ocr_results)

        partial_df.to_csv(
            OUTPUT_DIR / "test_ocr_partial.csv",
            index=False,
            encoding="utf-8-sig"
        )

        print(
            f"Đã xử lý tiếp {index}/{len(remaining_records)} "
            f"- Tổng kết quả: {len(test_ocr_results)}"
        )

# ==========================================
# 5. LƯU KẾT QUẢ OCR HOÀN CHỈNH
# ==========================================

test_ocr_df = pd.DataFrame(test_ocr_results)

test_ocr_df["key"] = test_ocr_df["key"].astype(str)

TEST_OCR_PATH = OUTPUT_DIR / "test_ocr.csv"

test_ocr_df.to_csv(
    TEST_OCR_PATH,
    index=False,
    encoding="utf-8-sig"
)

elapsed_time = time.time() - start_time

print("\n========== TEST OCR COMPLETED ==========")
print("Tổng số hóa đơn:", len(test_ocr_df))
print("Số mã duy nhất:", test_ocr_df["key"].nunique())
print("\nTrạng thái OCR:")
print(test_ocr_df["status"].value_counts(dropna=False))
print("Thời gian chạy tiếp:", round(elapsed_time, 2), "giây")
print("File đã lưu:", TEST_OCR_PATH)

# Kiểm tra
assert len(test_ocr_df) == 361
assert test_ocr_df["key"].nunique() == 361
assert set(test_ocr_df["split"].unique()) == {"test"}

print("\nTEST OCR DATA VALIDATION COMPLETED!")
```

```python

# ============================================================
# 1. CHUẨN BỊ VÀ KIỂM TRA DỮ LIỆU
# ============================================================

import pandas as pd
import numpy as np
import re
from collections import Counter
from datetime import datetime

FIELDS = ["company", "date", "address", "total"]

print("=" * 65)
print("BẮT ĐẦU ĐÁNH GIÁ TRÊN TEST SET")
print("=" * 65)

required_dfs = {
    "Ground Truth": gt_test,
    "Baseline": baseline_test_df,
    "Improved": improved_test_df
}

for name, df in required_dfs.items():

    if "key" not in df.columns:
        raise ValueError(f"{name} không có cột key!")

    missing = [f for f in FIELDS if f not in df.columns]

    if missing:
        raise ValueError(f"{name} thiếu cột: {missing}")

    if df["key"].duplicated().any():
        raise ValueError(f"{name} có key trùng!")

    print(f"{name}: {len(df)} hóa đơn")


# ============================================================
# 2. GHÉP DỮ LIỆU
# ============================================================

gt = gt_test[["key"] + FIELDS].rename(
    columns={f: f"{f}_gold" for f in FIELDS}
)

baseline = baseline_test_df[["key"] + FIELDS].rename(
    columns={f: f"{f}_baseline" for f in FIELDS}
)

improved = improved_test_df[["key"] + FIELDS].rename(
    columns={f: f"{f}_improved" for f in FIELDS}
)

data = gt.merge(baseline, on="key", how="inner")
data = data.merge(improved, on="key", how="inner")

if len(data) != len(gt_test):
    raise ValueError("Số hóa đơn sau khi ghép không khớp!")

total_invoices = len(data)

print("\nSố hóa đơn được đối chiếu:", total_invoices)


# ============================================================
# 3. HÀM CHUẨN HÓA
# ============================================================

def is_missing(value):

    if value is None:
        return True

    try:
        if pd.isna(value):
            return True
    except (TypeError, ValueError):
        pass

    return str(value).strip().lower() in ["", "nan", "none", "null"]


def norm_text(value):

    if is_missing(value):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(value).strip().lower()
    )


def norm_date(value):

    if is_missing(value):
        return ""

    core = re.sub(
        r"\s*([/.\-])\s*",
        r"\1",
        str(value).strip()
    )

    formats = [
        "%d/%m/%Y", "%m/%d/%Y", "%Y-%m-%d",
        "%d-%m-%Y", "%m-%d-%Y",
        "%Y/%m/%d", "%d.%m.%Y"
    ]

    for fmt in formats:

        try:
            return datetime.strptime(
                core,
                fmt
            ).strftime("%Y-%m-%d")

        except ValueError:
            continue

    return norm_text(core)


def norm_total(value):

    if is_missing(value):
        return ""

    text = str(value).strip()

    text = re.sub(
        r"(?i)\b(rm|vnd|usd|myr)\b",
        "",
        text
    )

    text = re.sub(r"\s+", "", text)

    # Xử lý dấu phân cách hàng nghìn và thập phân
    if "," in text and "." in text:

        if text.rfind(",") > text.rfind("."):
            text = text.replace(".", "").replace(",", ".")
        else:
            text = text.replace(",", "")

    elif "," in text:

        # Ví dụ 4,90 là số thập phân; 1,234 là hàng nghìn
        parts = text.split(",")

        if len(parts) == 2 and len(parts[1]) == 2:
            text = parts[0] + "." + parts[1]
        else:
            text = text.replace(",", "")

    match = re.search(r"-?\d+(?:\.\d+)?", text)

    if match:

        try:
            return f"{float(match.group()):.2f}"

        except ValueError:
            pass

    return text.lower()


def normalize_field(field, value):

    if field == "date":
        return norm_date(value)

    if field == "total":
        return norm_total(value)

    return norm_text(value)


# ============================================================
# 4. ĐÁNH GIÁ EXACT MATCH TỪNG TRƯỜNG
# ============================================================

metric_rows = []

for model in ["baseline", "improved"]:

    print(f"\nĐANG ĐÁNH GIÁ {model.upper()}")

    for field in FIELDS:

        gold_values = data[f"{field}_gold"].apply(
            lambda x: normalize_field(field, x)
        )

        pred_values = data[f"{field}_{model}"].apply(
            lambda x: normalize_field(field, x)
        )

        correct = gold_values == pred_values

        correct_count = int(correct.sum())

        accuracy = correct_count / total_invoices * 100

        metric_rows.append({
            "Model": model.capitalize(),
            "Field": field.capitalize(),
            "Correct": correct_count,
            "Total": total_invoices,
            "Exact Match (%)": round(accuracy, 2)
        })

        print(
            f"{field.capitalize():10s}: "
            f"{correct_count}/{total_invoices} = {accuracy:.2f}%"
        )

metrics_df = pd.DataFrame(metric_rows)


# ============================================================
# 5. ĐÁNH GIÁ TOKEN-LEVEL PRECISION, RECALL VÀ F1
#    Áp dụng cho COMPANY và ADDRESS
# ============================================================

token_metric_rows = []

TOKEN_FIELDS = ["company", "address"]

for model in ["baseline", "improved"]:

    for field in TOKEN_FIELDS:

        total_gold_tokens = 0
        total_pred_tokens = 0
        total_matched_tokens = 0

        for _, row in data.iterrows():

            # Chuẩn hóa văn bản trước khi tách token
            gold_text = norm_text(row[f"{field}_gold"])
            pred_text = norm_text(row[f"{field}_{model}"])

            # Tách token, không tính dấu câu là token
            gold_tokens = re.findall(
                r"[^\W_]+",
                gold_text,
                flags=re.UNICODE
            )

            pred_tokens = re.findall(
                r"[^\W_]+",
                pred_text,
                flags=re.UNICODE
            )

            # Dùng Counter để tính số token trùng,
            # có xét đến số lần xuất hiện của từng token
            gold_counter = Counter(gold_tokens)
            pred_counter = Counter(pred_tokens)

            matched_tokens = sum(
                (gold_counter & pred_counter).values()
            )

            total_gold_tokens += len(gold_tokens)
            total_pred_tokens += len(pred_tokens)
            total_matched_tokens += matched_tokens

        # Tính Precision, Recall và F1 theo tổng token
        precision = (
            total_matched_tokens / total_pred_tokens
            if total_pred_tokens > 0 else 0
        )

        recall = (
            total_matched_tokens / total_gold_tokens
            if total_gold_tokens > 0 else 0
        )

        f1 = (
            2 * precision * recall / (precision + recall)
            if precision + recall > 0 else 0
        )

        token_metric_rows.append({
            "Model": model.capitalize(),
            "Field": field.capitalize(),
            "Matched Tokens": total_matched_tokens,
            "Ground Truth Tokens": total_gold_tokens,
            "Predicted Tokens": total_pred_tokens,
            "Token Precision (%)": round(precision * 100, 2),
            "Token Recall (%)": round(recall * 100, 2),
            "Token F1 (%)": round(f1 * 100, 2)
        })

token_metrics_df = pd.DataFrame(token_metric_rows)

print("\n" + "=" * 65)
print("ĐÁNH GIÁ TOKEN-LEVEL PRECISION, RECALL VÀ F1")
print("=" * 65)

display(token_metrics_df)


# ============================================================
# 6. ĐÁNH GIÁ ĐỘ CHÍNH XÁC TOÀN BỘ HÓA ĐƠN
# ============================================================

invoice_results = data[["key"]].copy()

for model in ["baseline", "improved"]:

    matches = []

    for field in FIELDS:

        gold_values = data[f"{field}_gold"].apply(
            lambda x: normalize_field(field, x)
        )

        pred_values = data[f"{field}_{model}"].apply(
            lambda x: normalize_field(field, x)
        )

        correct = gold_values == pred_values

        invoice_results[f"{field}_{model}_correct"] = correct

        matches.append(correct)

    invoice_results[f"{model}_all_correct"] = (
        pd.concat(matches, axis=1).all(axis=1)
    )


baseline_correct = int(
    invoice_results["baseline_all_correct"].sum()
)

improved_correct = int(
    invoice_results["improved_all_correct"].sum()
)

baseline_accuracy = baseline_correct / total_invoices * 100
improved_accuracy = improved_correct / total_invoices * 100

print("\n" + "=" * 65)
print("ĐỘ CHÍNH XÁC TOÀN BỘ HÓA ĐƠN")
print("=" * 65)

print(
    f"Baseline: {baseline_correct}/{total_invoices} "
    f"= {baseline_accuracy:.2f}%"
)

print(
    f"Improved: {improved_correct}/{total_invoices} "
    f"= {improved_accuracy:.2f}%"
)


# ============================================================
# 7. BẢNG SO SÁNH HAI MÔ HÌNH
# ============================================================

comparison_rows = []

for field in FIELDS:

    base = metrics_df[
        (metrics_df["Model"] == "Baseline") &
        (metrics_df["Field"] == field.capitalize())
    ]["Exact Match (%)"].iloc[0]

    imp = metrics_df[
        (metrics_df["Model"] == "Improved") &
        (metrics_df["Field"] == field.capitalize())
    ]["Exact Match (%)"].iloc[0]

    comparison_rows.append({
        "Field": field.capitalize(),
        "Baseline (%)": base,
        "Improved (%)": imp,
        "Difference (percentage points)": round(imp - base, 2)
    })


comparison_rows.append({
    "Field": "All Invoice",
    "Baseline (%)": round(baseline_accuracy, 2),
    "Improved (%)": round(improved_accuracy, 2),
    "Difference (percentage points)": round(
        improved_accuracy - baseline_accuracy,
        2
    )
})

comparison_df = pd.DataFrame(comparison_rows)

print("\nBẢNG SO SÁNH BASELINE VÀ IMPROVED")

display(comparison_df)


# ============================================================
# 8. PHÂN TÍCH LỖI
# ============================================================

error_rows = []

for model in ["baseline", "improved"]:

    for _, row in data.iterrows():

        for field in FIELDS:

            gold_value = normalize_field(
                field,
                row[f"{field}_gold"]
            )

            pred_value = normalize_field(
                field,
                row[f"{field}_{model}"]
            )

            if gold_value != pred_value:

                error_rows.append({
                    "key": row["key"],
                    "Model": model.capitalize(),
                    "Field": field.capitalize(),
                    "Ground Truth": row[f"{field}_gold"],
                    "Prediction": row[f"{field}_{model}"]
                })


error_df = pd.DataFrame(
    error_rows,
    columns=[
        "key",
        "Model",
        "Field",
        "Ground Truth",
        "Prediction"
    ]
)

print("\nPHÂN TÍCH LỖI DỰ ĐOÁN")

print(
    "Số lỗi Baseline:",
    len(error_df[error_df["Model"] == "Baseline"])
)

print(
    "Số lỗi Improved:",
    len(error_df[error_df["Model"] == "Improved"])
)

display(error_df.head(20))


# ============================================================
# 9. NHẬN XÉT VÀ KẾT LUẬN TỰ ĐỘNG
# ============================================================

print("\n" + "=" * 65)
print("NHẬN XÉT VÀ ĐÁNH GIÁ KẾT QUẢ")
print("=" * 65)

print(f"""
1. ĐÁNH GIÁ CHUNG

- Bộ Test gồm {total_invoices} hóa đơn.
- Baseline trích xuất chính xác toàn bộ 4 trường trên
  {baseline_correct}/{total_invoices} hóa đơn ({baseline_accuracy:.2f}%).
- Improved trích xuất chính xác toàn bộ 4 trường trên
  {improved_correct}/{total_invoices} hóa đơn ({improved_accuracy:.2f}%).
""")

print("2. SO SÁNH ĐỘ CHÍNH XÁC TỪNG TRƯỜNG")

for field in FIELDS:

    base = metrics_df[
        (metrics_df["Model"] == "Baseline") &
        (metrics_df["Field"] == field.capitalize())
    ]["Exact Match (%)"].iloc[0]

    imp = metrics_df[
        (metrics_df["Model"] == "Improved") &
        (metrics_df["Field"] == field.capitalize())
    ]["Exact Match (%)"].iloc[0]

    difference = imp - base

    print(
        f"- {field.capitalize()}: "
        f"Baseline {base:.2f}%, Improved {imp:.2f}%."
    )

    if difference > 0:
        print(
            f"  Improved tăng {difference:.2f} điểm phần trăm."
        )

    elif difference < 0:
        print(
            f"  Improved giảm {abs(difference):.2f} điểm phần trăm."
        )

    else:
        print("  Hai mô hình có kết quả tương đương.")


print("\n3. SO SÁNH TOKEN-LEVEL F1")

for field in TOKEN_FIELDS:

    base_f1 = token_metrics_df[
        (token_metrics_df["Model"] == "Baseline") &
        (token_metrics_df["Field"] == field.capitalize())
    ]["Token F1 (%)"].iloc[0]

    imp_f1 = token_metrics_df[
        (token_metrics_df["Model"] == "Improved") &
        (token_metrics_df["Field"] == field.capitalize())
    ]["Token F1 (%)"].iloc[0]

    difference = imp_f1 - base_f1

    print(
        f"- {field.capitalize()}: "
        f"Baseline F1 = {base_f1:.2f}%, "
        f"Improved F1 = {imp_f1:.2f}%."
    )

    if difference > 0:
        print(f"  Improved tăng {difference:.2f} điểm phần trăm F1.")

    elif difference < 0:
        print(f"  Improved giảm {abs(difference):.2f} điểm phần trăm F1.")

    else:
        print("  Hai mô hình có Token-level F1 bằng nhau.")


print("\n4. SO SÁNH TỔNG THỂ")

if improved_accuracy > baseline_accuracy:

    print(
        f"- Improved tốt hơn Baseline "
        f"{improved_accuracy - baseline_accuracy:.2f} "
        "điểm phần trăm về độ chính xác toàn bộ hóa đơn."
    )

elif improved_accuracy < baseline_accuracy:

    print(
        f"- Improved thấp hơn Baseline "
        f"{baseline_accuracy - improved_accuracy:.2f} "
        "điểm phần trăm về độ chính xác toàn bộ hóa đơn."
    )

else:
    print("- Hai mô hình có độ chính xác toàn bộ hóa đơn bằng nhau.")


print("""
5. NHẬN XÉT VỀ HẠN CHẾ

- Exact Match yêu cầu giá trị dự đoán phải trùng khớp
  sau khi chuẩn hóa với Ground Truth.
- Token-level F1 đánh giá mức độ trùng khớp giữa các token,
  giúp phản ánh mức độ chính xác một phần của Company và Address.
- OCR có thể nhận dạng sai ký tự, chữ số, dấu câu hoặc bố cục.
- Sai sót ở một trường có thể khiến cả hóa đơn bị tính là không chính xác.
- Kết quả của Improved cần được đánh giá riêng trên từng trường,
  không mặc định mô hình chỉ cải thiện hoặc làm giảm một trường cụ thể.

6. KẾT LUẬN

Kết quả Test set giúp so sánh khả năng trích xuất thông tin
của Baseline và Improved trên dữ liệu chưa sử dụng trong
quá trình xây dựng mô hình.

Việc kết hợp Exact Match và Token-level F1 giúp đánh giá
cả độ chính xác tuyệt đối và mức độ trùng khớp từng phần
của các trường Company và Address.
""")


# ============================================================
# 10. LƯU TẤT CẢ KẾT QUẢ
# ============================================================

OUTPUT_DIR = PROJECT_DIR / "outputs" / "test"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

metrics_df.to_csv(
    OUTPUT_DIR / "test_metrics_summary.csv",
    index=False,
    encoding="utf-8-sig"
)

token_metrics_df.to_csv(
    OUTPUT_DIR / "test_token_metrics.csv",
    index=False,
    encoding="utf-8-sig"
)

invoice_results.to_csv(
    OUTPUT_DIR / "test_invoice_accuracy.csv",
    index=False,
    encoding="utf-8-sig"
)

comparison_df.to_csv(
    OUTPUT_DIR / "test_comparison.csv",
    index=False,
    encoding="utf-8-sig"
)

error_df.to_csv(
    OUTPUT_DIR / "test_error_analysis.csv",
    index=False,
    encoding="utf-8-sig"
)

data.to_csv(
    OUTPUT_DIR / "test_full_comparison.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 65)
print("TEST EVALUATION COMPLETED!")
print("=" * 65)

print("Đã lưu 6 file kết quả trong:", OUTPUT_DIR)
print("\nDanh sách file:")

for file_path in sorted(OUTPUT_DIR.glob("*.csv")):
    print("-", file_path.name)
```
